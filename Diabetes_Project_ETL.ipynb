```python
import matplotlib.pyplot as plt

# Data
categories = ['No Diabetes', 'Diabetes']
bmi_values = [27.88, 31.96]

# Create chart
plt.figure(figsize=(6,4))
plt.bar(categories, bmi_values)

# Labels and title
plt.title('Average BMI by Diabetes Status')
plt.ylabel('Average BMI')
plt.xlabel('Diabetes Status')

# Show chart
plt.show()
```

```python
import matplotlib.pyplot as plt

categories = ['No Diabetes', 'Diabetes']
bmi_values = [27.88, 31.96]

plt.figure(figsize=(6,4))
bars = plt.bar(categories, bmi_values)

plt.title('Average BMI by Diabetes Status')
plt.ylabel('Average BMI')
plt.xlabel('Diabetes Status')

# Add labels on bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,
             height + 0.3,
             round(height,2),
             ha='center')

plt.show()
```

```python
# Pie chart for diabetes distribution

labels = ['No Diabetes', 'Diabetes']
sizes = [df_clean['Diabetes_binary'].value_counts()[0],
         df_clean['Diabetes_binary'].value_counts()[1]]

plt.figure(figsize=(6,6))
plt.pie(sizes, labels=labels, autopct='%1.1f%%')

plt.title('Distribution of Diabetes Status')

plt.show()
```
